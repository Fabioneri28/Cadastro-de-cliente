```python
df['senioridade'] = df['senioridade'].replace({
    'SE': 'Senior',
    'MI': 'Pleno',
    'EN': 'Júnior',
    'EX': 'Executivo'
})

print("Categorias da coluna 'senioridade' traduzidas com sucesso:")
print(df['senioridade'].value_counts())
```

```python
df['remoto'] = df['remoto'].replace({
    0: 'Presencial',
    100: 'Remoto',
    50: 'Híbrido'
})

print("Categorias da coluna 'remoto' traduzidas com sucesso:")
print(df['remoto'].value_counts())
```

```python
import numpy as np

df_salarios = pd.DataFrame({
    "nome":["Ana","Bruno","carlos","Daniele","Val","Fabio", "Kelly", "Vinicius", "Manoella"],
    "salarios":[4000,np.nan,5000,np.nan,10000,2500,3500,12500,7780]
})
#callcula a media de salarios e sibstitui pela media earredondamento
df_salarios["salarios_media"] = df_salarios["salarios"].fillna(round(df_salarios["salarios"].mean(), 2))
'''Se você quiser preencher com a mediana em uma nova coluna, use a linha abaixo, ou remova se a média for suficiente.

lembarnod que cada df_... de analise deverá obter as mesma nomeações na base para que hajá a leitura ideal
'''
df_salarios["salarios_mediana"] = df_salarios["salarios"].fillna(round(df_salarios["salarios"].median()))
#impirmi os df_salarios
df_salarios
```

```python
import sys
!{sys.executable} -m pip install pycountry
import plotly.express as px
import pandas as pd
import pycountry

# Load df_limpo from the saved CSV to ensure it's available
df_limpo = pd.read_csv('dados-imersao-final.csv')

# Function to get ISO-3 country code
def get_iso3(country_code):
    try:
        return pycountry.countries.get(alpha_2=country_code).alpha_3
    except AttributeError:
        return None

# Create 'residencia_iso3' column
df_limpo['residencia_iso3'] = df_limpo['residencia'].apply(get_iso3)

# Filter out rows where ISO-3 code couldn't be found (optional, but good practice for maps)
df_limpo_map = df_limpo.dropna(subset=['residencia_iso3'])

# Calcular média salarial por país (ISO-3)
df_ds = df_limpo_map[df_limpo_map['cargo'] == 'Data Scientist']
media_ds_pais = df_ds.groupby('residencia_iso3')['usd'].mean().reset_index()

# Gerar o mapa
fig = px.choropleth(media_ds_pais,
                    locations='residencia_iso3',
                    color='usd',
                    color_continuous_scale='rdylgn',
                    title='Salário médio de Cientista de Dados por país',
                    labels={'usd': 'Salário médio (USD)', 'residencia_iso3': 'País'})

fig.show()
```
